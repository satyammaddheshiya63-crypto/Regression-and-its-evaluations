**Regressin and Its Evaluation**


1 : What is simple Linear Regression?

Ans : Simple Linear Regression is a basic statistical method used to model the relationship between two variables.

One independent variable : input or predictor
one dependent variable : output or response

2 : What are the key assumptions of Simple Linear Regression?

Ans : Key Assumption of Simple Linear Regression : The relationship between X and Y should be linear Or a straight line.

Independece of Errors : The errors or residuals should be independent of each other.

HomoScedasticity(Constant Variance) : The spread of errors should be constant across all values of X.

3. What is heteroscedasticity, and why is it important to address in regression model?

Ans :  It means that the variance of the errors is not constant across all values of the independent variable(X).

As X changes, The errors don't stay evenly spread- they may increase or decrease, often forming a funnel shape.

X = Income
Y - Spending

Heteroscedasticity violates one of the key assumptions of regression (constant variance). if you ignore it

: Unreliable Predictions
Wrong Standard Errors
Invalid Hypothesis Testing
Misleading feature importance


4 : What is the multiple linear regression?

Ans : It is an extension of simple linear regression where we use two or more independent variables (X1, X2, ....) to predict a dependent variable (Y).

Y = a + b1X1 + b2X2 + ..... + bnXn

Y = dependent variable(output)
X1, X2 .... = Independent variables (inputs)
a = intercept
b1, b2 .... = coefficients


5 : What is polynomial regression, and how does it differ from linear regression?

Ans : It is a type of regression where the relationship between the input(X) and output(Y) is modeled using higher powers of X, instead of just a straight line.

Feature     Linear Regression     Polynomial Regression
Shape       Straight line         Curved line
Equation    Y = a + bX            Y = a+b1X + b2X
Flexibility   Low                 High
Use Case      Linear Regression   Non-linear Regression
